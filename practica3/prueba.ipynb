```python
def expmod(a,n,z):
    i = n;x = a;r = 1
    while i > 0:
        if i % 2 == 1:
            r = (r * x ) % z
            #print(r,i)
            i -= 1 
        x = (x * x ) % z
        i = i/2
    return r
```

```python
from math import isqrt
from collections import defaultdict

def baby_step_giant_step(g, h, p):
    m = isqrt(p - 1) + 1

    # Baby steps: g^j mod p
    baby_steps = {}
    for j in range(m):
        value = pow(g, j, p)
        baby_steps[value] = j

    # Compute g^-m mod p
    g_inv_m = pow(g, -m, p)

    # Giant steps: h * g^(-im)
    gamma = h
    for i in range(m):
        if gamma in baby_steps:
            return i * m + baby_steps[gamma]
        gamma = (gamma * g_inv_m) % p

    return None  # No solution found

# Ejemplo
g = 2
h = 7
p = 15
x = baby_step_giant_step(g, h, p)
print(f"Solución: x = {x} → {g}^{x} ≡ {h} mod {p}")
```
