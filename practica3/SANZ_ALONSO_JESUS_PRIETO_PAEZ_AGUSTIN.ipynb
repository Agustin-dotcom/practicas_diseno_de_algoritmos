Exponenciación modular

```python
def expmod(a,n,z):
    i = n;x = a;r = 1
    while i > 0:
        if i % 2 == 1:
            r = (r * x ) % z
            #print(r,i)
            i -= 1
        x = (x * x ) % z
        i = i/2
    return r
```

```python
from math import isqrt
from collections import defaultdict

def modinv(a, p):
    # Inverso modular usando el algoritmo extendido de Euclides
    return pow(a, -1, p)

def baby_step_giant_step(g, h, p):
    m = isqrt(p) + 1

    # Baby steps
    baby = {}
    for j in range(m):
        baby[pow(g, j, p)] = j

    # g^-m mod p
    g_inv_m = pow(modinv(g, p), m, p)

    # Giant steps
    gamma = h
    for i in range(m):
        if gamma in baby:
            return i * m + baby[gamma]
        gamma = gamma * g_inv_m % p

    return None  # No solución encontrada

# Valores dados
g = 12345709
h =  379615130#575137720#379615130
p = 666666667

x = baby_step_giant_step(g, h, p)

if x is not None:
    print(f"✅ Solución encontrada: x = {x}")
    # Verificación:
    assert pow(g, x, p) == h, "❌ Verificación fallida"
else:
    print("❌ No se encontró solución")
```
