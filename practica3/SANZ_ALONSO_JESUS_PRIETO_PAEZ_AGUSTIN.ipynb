```python
num1 = int('8'*1000)
num2 = int('9'*1000)
```

```python

def multi(u,v):
    #global n
    #global s;global x;global y;global r;global p;global q; global diez_a_la_s
    n,s = 0,0;w,x,y,r,p,q=0,0,0,0,0,0
    tam_u = len(str(int(u)))
    #print('tam_u ',tam_u)
    tam_v = len(str(int(v)))
    #print('tam_v ',tam_v)
    n = np.max([tam_u,tam_v])
    #print(n)
    #print(type(n))
    #print('n ',n)
    if n == 1:
        return u*v
    else:
        s = int(n/2)
        #print(str(type(s)))
        #print(10**s)
        diez_a_la_s = 10**s
        #print(diez_a_la_s)
        w = u//diez_a_la_s;x = u % diez_a_la_s
        y = v//diez_a_la_s;z = v % diez_a_la_s
        #print('w+x ',w+x)
        #print('y+z ',y+z )
        r = multi(w+x,y+z)
        p = multi(w,y)
        q = multi(x,z)
        return p * (10**(2*s)) + (r-p-q) * diez_a_la_s + q
t = time()*1000
print(multi(num1,num2))
t = time()*1000 - t
print('Tiempo empleado ',t,' ms')
```

```python
def expmod(a,n,z):
    i,x,r = 0,0,0
    i = n;x = a;r = 1
    while i > 0:
        if i % 2 != 0:
            r = multi(r , x % z)
        x = multi(x , x % z)
        i = i/2
    return r
```

```python
t = time()*1000
print(570**670 % 23)
t = time()*1000-t
print('Tiempo empleado ',t,' ms')
t = time()*1000
print(expmod(570,670,23))
t = time()*1000 -t
print('Tiempo empleado ',t,' ms')
```
