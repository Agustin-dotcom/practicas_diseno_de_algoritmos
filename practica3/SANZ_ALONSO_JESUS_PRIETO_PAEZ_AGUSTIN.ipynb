```python
def multi(u,v):
    #global n
    #global s;global x;global y;global r;global p;global q; global diez_a_la_s
    n,s = 0,0;w,x,y,r,p,q=0,0,0,0,0,0
    tam_u = len(str(int(u)))
    #print('tam_u ',tam_u)
    tam_v = len(str(int(v)))
    #print('tam_v ',tam_v)
    n = np.max([tam_u,tam_v])
    #print(n)
    #print(type(n))
    #print('n ',n)
    if n == 1:
        return u*v
    else:
        s = int(n/2)
        #print(str(type(s)))
        #print(10**s)
        diez_a_la_s = 10**s
        #print(diez_a_la_s)
        w = u//diez_a_la_s;x = u % diez_a_la_s
        y = v//diez_a_la_s;z = v % diez_a_la_s
        #print('w+x ',w+x)
        #print('y+z ',y+z )
        r = multi(w+x,y+z)
        p = multi(w,y)
        q = multi(x,z)
        return p * (10**(2*s)) + (r-p-q) * diez_a_la_s + q
t = time()*1000
print(multi(num1,num2))
t = time()*1000 - t
print('Tiempo empleado ',t,' ms')
```

```python
def karatsuba(u, v):
    if u < 10 or v < 10:
        return u * v

    s = max(len(str(u)), len(str(v)))
    half_s = s // 2

    power_of_10 = 10**half_s

    w = u // power_of_10
    x = u % power_of_10
    y = v // power_of_10
    z = v % power_of_10

    r = karatsuba(w + x, y + z)
    p = karatsuba(w, y)
    q = karatsuba(x, z)

    return p * (10**(2 * half_s)) + (r - p - q) * power_of_10 + q

from time import time

num1 = 12345678901234567890
num2 = 98765432109876543210

t = time() * 1000
resultado = karatsuba(num1, num2)
t = time() * 1000 - t
print(resultado)
print('Tiempo empleado ', t, ' ms')

t_std = time() * 1000
resultado_std = num1 * num2
t_std = time() * 1000 - t_std
print(resultado_std)
print('Tiempo empleado (estándar) ', t_std, ' ms')
```

```python
def expmod(a,n,z):
    i = n;x = a;r = 1
    while i > 0:
        if i % 2 == 1:
            r = karatsuba(r , x ) % z
            #print(r,i)
        x = karatsuba(x , x ) % z
        i = i//2
    return r
```
