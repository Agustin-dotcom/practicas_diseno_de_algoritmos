```python
##módulo (número primo) (p): 23 (En el enunciado de la práctica, a p se le llama n)
##base (g): 5
## y g lo saben los dos comunicantes y todos

## a=6
##Solo A lo sabe

##A envia el mensaje: g^a mod p = 5^6 mod 23 = 8 (z1)
##B recibe el mensaje: z1^b mod p = 8^15 mod 23 = 2 


##b=15
##Solo B lo sabe

##B envia el mensaje: g^b mod p = 5^15 mod 23 = 19 (z2)
##A recibe el mensaje: z2^a mod p = 19^6 mod 23 = 2

##Cuando reciben el mensaje A y B y hacen los cálculos, sale el mismo número. Esto solo puede pasar si A sabe a y B sabe b.

##Un hacker tendría que probar todos los números de 1 a p-1 (de 1 a 22, estos incluídos) para resolver las siguientes ecuaciones:

##5^x mod 23 = 8. En este caso, x debería dar 6 (Del mensaje que envía A)

##5^y mod 23 = 19. En este caso, y debería dar 15 (Del mensaje que envía B)

##No hace falta hallar x para hallar y, y viceversa. Se pueden conseguir independientemente.

##Si tienes un número y, para cada iteracción, lo multiplicas por c y le haces el módulo de d, al final el resultado es el mismo que si multiplicas por c en todas
##las iteracciones y solo al final aplicas el módulo d. Esta igualdad nos va a permitir trabajar mucho con los números sin la necesidad de tener que tratar con 
##números tan grandes
```

```python
p=23
g=5  ##p y g son los valores del módulo y la base (lo saben todos)

z1=8
z2=19 ##z1 y z2 son los valores del mensaje (lo saben todos)

i1=1 ##Uso el mismo contador para las 2 incógnitas

x=1
y=1 ##x e y son los valores que estamos calculando para decodificar, respectivamente, los mensajes z1 y z2

##Decodificar z1 hallando x: g^x mod p = z1
##Decodificar z2 hallando y: g^y mod p = z2
def ExpMod (a, n, z): ##este al final no lo hemos usado
    r=1
    if n==0:
        return 1
    else:
        if n%2!=0:
            r=ExpMod(a, n-1, z)
            return a*r % z
        else:
            r=ExpMod(a, n/2, z)
            return r*r % z

print(ExpMod(5, 6, 23))

def ExpMod2 (a, n, z): ##ESTE ES EL QUE HEMOS USADO
    i=n
    x=a
    r=1
    while i>0:
        if i%2!=0:
            r=r*x % z
            i=i-1
        x=x*x % z
        i=i/2
    return r
print(ExpMod2(5, 6, 23)) ##¿Está mal?
```

```python
p=23
g=5  ##p y g son los valores del módulo y la base (lo saben todos)

z1=8
z2=19 ##z1 y z2 son los valores del mensaje (lo saben todos)

i1=1 ##Uso la misma variable para las 2 incógnitas

x=0
y=0 ##x e y son los valores que estamos calculando para decodificar, respectivamente, los mensajes z1 y z2

##Decodificar z1 hallando x: g^x mod p = z1
##Decodificar z2 hallando y: g^y mod p = z2
print("g: ",g,". p: ",p,". z1=",z1,". z2=",z2)
print("Hallando x e y para decodificar z1 y z2") ##En cada iteracción, miro si el valor coincide con alguno de los valores de los mensajes. 
while i1 != z1 and i1 != z2 and x<p: ## and x (o y)<p
    x=x+1
    y=y+1
    i1=ExpMod2(g, x, p) ##También podría ser ExpMod(g, y, p)
    print("x: ",x,". g^x % p: ",i1,". z1: ",z1,". y: ",y,". g^y % p: ",i1,". z2: ",z2)
if i1==z1: ##Si he hallado x, sigo buscando y, pero ya he terminado con x
    print("SOLUCION: ")
    print("x: ",x,". g^x % p: ",i1,". z1: ",z1)
    while i1!=z2 and y<p:
        y=y+1
        i1=ExpMod2(g, y, p)
        print("y: ",y,". g^y % p: ",i1,". z2: ",z2)
    print("SOLUCION: ")
    print("y: ",y,". g^y % p: ",i1,". z2: ",z2)
else: ##Si llego aquí, es porque no he hallado x, pero sí he hallado y. Si he hallado y, sigo buscando x, pero ya he terminado con y.
    print("SOLUCION: ")
    print("y: ",y,". g^y % p: ",i1,". z2: ",z2)

    while i1!=z1 and x<p:
        x=x+1
        i1=ExpMod2(g, x, p)
        print("x: ",x,". g^x % p: ",i1,". z1: ",z1)
    print("SOLUCION: ")
    print("x: ",x,". g^x % p: ",i1,". z1: ",z1)
print("x es ",x,". y es ",y)
```
